```python
import pandas as pd
import os
```

```python
new_profiles = {}

new_subregion_profile_path = '../eia_data_cleaner/data/subregional/outputs'
for fname in os.listdir(new_subregion_profile_path):
    fpath = os.path.join(new_subregion_profile_path, fname)
    if os.path.isdir(fpath):
        continue
    df = pd.read_csv(fpath, parse_dates=["date_time"], index_col="date_time")
    subregion = fname.replace('.csv', '')
    new_profiles[subregion] = df

new_region_profile_path = '../eia_data_cleaner/data/regional/outputs'
for fname in os.listdir(new_region_profile_path):
    fpath = os.path.join(new_region_profile_path, fname)
    if os.path.isdir(fpath):
        continue
    df = pd.read_csv(fpath, parse_dates=["date_time"], index_col="date_time")
    region = fname.replace('.csv', '')
    new_profiles[region] = df

new_region_profile_path = '../eia_data_cleaner/data/forecast/outputs'
for fname in os.listdir(new_region_profile_path):
    fpath = os.path.join(new_region_profile_path, fname)
    if os.path.isdir(fpath):
        continue
    df = pd.read_csv(fpath, parse_dates=["date_time"], index_col="date_time")
    region = fname.replace('.csv', '')
    new_profiles[f"{region}_forecast"] = df
```

```python
new_profiles_adj = {}
for respondent, df in new_profiles.items():
    first_index = df.loc[~df['raw demand (MW)'].isna()].index[0]
    last_index = df.loc[~df['raw demand (MW)'].isna()].index[-1]
    if (
        (not ((first_index.year == 2016) and (first_index.month == 1))) |
        (not ((last_index.year == 2023) and (last_index.month == 12)))
    ):
        print(respondent)
        new_profiles_adj[respondent] = df.loc[(df.index >= first_index) & (df.index <= last_index)]
    else:
        new_profiles_adj[respondent] = new_profiles[respondent]
```

```python
changes = {}
lengths = {}
for respondent, df in new_profiles_adj.items():
    changes[respondent] = len(df.loc[df.category != 'OKAY'])
    lengths[respondent] = len(df)
```

```python
sum(changes.values()) / sum(lengths.values())
```
