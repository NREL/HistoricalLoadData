```python
import pandas as pd
import os
```

```python
# Profiles downloaded from https://portal.spp.org/pages/hourly-load
load_profile_list = []

dir = '../data/iso_load_profiles/raw/spp'
for filename in os.listdir(dir):
    f = os.path.join(dir, filename)
    df = pd.read_csv(f)
    load_profile_list.append(df)
```

```python
spp_load = pd.concat(load_profile_list)
spp_load['timestamp'] = pd.to_datetime(spp_load["MarketHour"], format='mixed')
spp_load = pd.melt(
    spp_load,
    id_vars=['MarketHour', 'timestamp'],
    value_vars=spp_load.drop(columns=['MarketHour', 'timestamp']).columns,
    var_name='subba'
)
spp_load['subba'] = spp_load['subba'].str.strip()
spp_load = (
    spp_load.dropna(subset='timestamp')
    .sort_values("timestamp")
    .reset_index(drop=True)
    [['timestamp', 'subba', 'value']]
)
```

```python
spp_load.to_csv(f"../data/iso_load_profiles/spp.csv", index=False)
```
