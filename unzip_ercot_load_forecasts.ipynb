```python
import pandas as pd
import os
import zipfile
```

```python
source_dir = '../../Downloads'
temp_destination_dir = 'data/iso_load_profiles/raw/ercot_forecast_temp'
destination_dir = 'data/iso_load_profiles/raw/ercot_forecast'

os.makedirs(temp_destination_dir, exist_ok=True)
for dir in os.listdir(source_dir):
    if ('cdr.np3-565' in dir) and (dir.endswith('.zip')):
        zip_dir = os.path.join(source_dir, dir)
        with zipfile.ZipFile(zip_dir, 'r') as zip_ref:
            zip_ref.extractall(temp_destination_dir)

os.makedirs(destination_dir, exist_ok=True)
for dir in os.listdir(temp_destination_dir):
    zip_dir = os.path.join(temp_destination_dir, dir)
    with zipfile.ZipFile(zip_dir, 'r') as zip_ref:
        zip_ref.extractall(destination_dir)
```
